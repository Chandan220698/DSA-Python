## Find the second max number

```python
nums = [1, 2, -10, 100, 90, 6, -16]
```

```python
def secondMax(nums: list):
    h1 = nums[0]
    h2 = nums[1]
    if h1 < h2:
        h1, h2 = h2, h1
    i = 2
    while i < len(nums):
        if nums[i] > h2:
            h2 = nums[i]
            if h1 < h2:
                h1, h2 = h2, h1
        i += 1
    return h1, h2
```

```python
secondMax(nums)
```

```python
secondMax([-56, 1])
```

```python
secondMax([-56, 1, 2, 3, 5, 5, 5])
```

```python
def secondMax(nums: list):
    h1 = float('-inf')
    h2 = float('-inf')
    for i in range(0, len(nums)):
        if nums[i] > h1:
            h2 = h1
            h1 = nums[i]
        elif nums[i] > h2 and nums[i] != h1:
            h2 = nums[i]
    return h1, h2
```

## Leetcode 283: Move Zeros to the End of the List

```python
def moveZeros(nums: list):
    p_zero = 0
    n = len(nums)
    for i in range(1, n):
        if nums[p_zero] != 0:
            p_zero += 1
        else:
            if nums[i] != 0:
                temp = nums[i]
                nums[i] = nums[p_zero]
                nums[p_zero] = temp
                p_zero += 1
    return nums
```

```python
nums = [1,1,0,3,12]
```

```python
moveZeros(nums)
```

```python
moveZeros([1,2,3,4,5])
```

```python
moveZeros([0])
```

## Merge 2 Sorted Arrays Without Duplicates

```python
# Union of Two Sorted Arrays with Distinct Elements
def findUnion(a: list[int], b: list[int]) -> list[int]:
    m = len(a)
    n = len(b)
    i, j = 0, 0
    prev = None
    output = []
    while i < m and j < n:
        if a[i] <= b[j]:
            if prev != a[i]:
                output.append(a[i])
                prev = a[i]
            i += 1
        else:
            if prev != b[j]:
                output.append(b[j])
                prev = b[j]
            j += 1

    while i < m:
        if prev != a[i]:
            output.append(a[i])
            prev = a[i]
        i += 1
    while j < n:
        if prev != b[j]:
            output.append(b[j])
            prev = b[j]
        j += 1
        
    return output
```

```python
a = [1, 2, 3, 4, 5]
b = [1, 2, 3, 6, 7]
# Output = [1, 2, 3, 4, 5, 6, 7]
```

```python
findUnion(a, b)
```

## Leetcode 268: Find Missing Number in an Array

```python
def missingNumber(nums):
    total = 0
    n = len(nums)
    for i in range(n):
        total += nums[i]
    return (((n+1)*n)//2) - total
```

```python
nums = [3,0,1]
missingNumber(nums)
```
