## Leetcode 73: Set Matrix Zeros

```python
"""
Set Matrix Zeroes
Medium
Array
Hash Table
Matrix
Given an m x n integer matrix matrix, if an element is 0, set its entire row and column to 0's.

You must do it in place.

Example 1:

Input: matrix = [[1,1,1],[1,0,1],[1,1,1]]
Output: [[1,0,1],[0,0,0],[1,0,1]]
Explanation: The element at position (1,1) is 0, so the entire row 1 and column 1 are set to 0.
Example 2:

Input: matrix = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]
Output: [[0,0,0,0],[0,4,5,0],[0,3,1,0]]
Explanation: The elements at positions (0,0) and (0,3) are 0, so rows and columns containing those zeroes are set to 0.
Example 3:

Input: matrix = [[1,2,3],[4,0,6],[7,8,9]]
Output: [[1,0,3],[0,0,0],[7,0,9]]
Explanation: The element at position (1,1) is 0, so the entire row 1 and column 1 are set to 0.
Constraints:

•
m == matrix.length
•
n == matrix[0].length
•
1 <= m, n <= 200
•
-231 <= matrix[i][j] <= 231 - 1
Input Format

matrix: List[List[int]] - An m x n integer matrix

Output Format

None - Modify the matrix in-place, do not return anything.
"""
```

```python
def setZeroes(matrix: list[list[int]]):
    m = len(matrix)
    n = len(matrix[0])
    row = [0 for _ in range(m)]
    col = [0 for _ in range(n)]

    for i in range(m):
        for j in range(n):
            if matrix[i][j] == 0:
                row[i] = -1
                col[j] = -1

    for i in range(m):
        for j in range(n):
            if (row[i] == -1) or (col[j] == -1):
                matrix[i][j] = 0
```

```python
matrix = [[1,1,1],[1,0,1],[1,1,1]]
print(matrix)
setZeroes(matrix)
matrix
```

```python
matrix = [[1, 0, 2, 3], [4, 5, 8, 0], [0, 9, 4, 5], [4, 9, 4, 5]]
print(matrix)
setZeroes(matrix)
matrix
```

## Leetcode 48: Rotate Matrix by 90 Degrees

```python
"""
Rotate Image
Medium
Array
Math
Matrix
You are given an n x n 2D matrix representing an image, rotate the image by 90 degrees (clockwise).

You have to rotate the image in-place, which means you have to modify the input 2D matrix directly. DO NOT allocate another 2D matrix and do the rotation.

Example 1:

Input: matrix = [[1,2,3],[4,5,6],[7,8,9]]
Output: [[7,4,1],[8,5,2],[9,6,3]]
Explanation: The matrix is rotated 90 degrees clockwise.
Example 2:

Input: matrix = [[5,1,9,11],[2,4,8,10],[13,3,6,7],[15,14,12,16]]
Output: [[15,13,2,5],[14,3,4,1],[12,6,8,9],[16,7,10,11]]
Explanation: The 4x4 matrix is rotated 90 degrees clockwise.
Example 3:

Input: matrix = [[1]]
Output: [[1]]
Explanation: A 1x1 matrix remains the same after rotation.
Constraints:

•
n == matrix.length == matrix[i].length
•
1 <= n <= 20
•
-1000 <= matrix[i][j] <= 1000
Input Format

matrix: List[List[int]] - An n x n 2D matrix representing an image

Output Format

None - Modify the matrix in-place, do not return anything
"""
```

```python
Input: matrix = [[1,2,3],[4,5,6],[7,8,9]]
Output: [[7,4,1],[8,5,2],[9,6,3]]
```

```python
def rotate(matrix: list[list[int]]) -> None:
    # Transpose the matrix then mirror (reverse each row)
    m = len(matrix)
    n = len(matrix[0])
    # Transpose the matrix
    for i in range(0, m):
        for j in range(i+1, n):    ## <--- logic
            temp = matrix[i][j]
            matrix[i][j] = matrix[j][i]
            matrix[j][i] = temp
    # Mirror, Reverse every rows of matrix
    for i in range(m):
        # m[i] -> row
        row = matrix[i]
        for j in range(n//2):      ## <--- logic
            temp = row[j]
            row[j] = row[n-j-1]
            row[n-j-1] = temp
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
rotate(matrix)
matrix
```

```python
matrix = [[5,1,9,11],[2,4,8,10],[13,3,6,7],[15,14,12,16]]
rotate(matrix)
matrix
```
