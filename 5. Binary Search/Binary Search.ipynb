## Binary Search

```python
def binarySearch(nums: list[int], target: int) -> int:
    l = 0
    h = len(nums) - 1
    while l <= h:
        mid = (l + h)//2
        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            h = mid - 1
        else:
            l = mid + 1
    return -1
```

```python
nums = [-1,0,3,5,9,12]
target = 9
binarySearch(nums, target)
```

```python
nums = [-1,0,3,5,9,12]
target = 2
binarySearch(nums, target)
```

```python
nums = [1,3,5,7,9]
target = 5
binarySearch(nums, target)
```

```python
nums = []
target = 5
binarySearch(nums, target)
```

#### Recursive Approch

```python
def binarySearchRecursive(nums: list[int], target: int, l: int, h: int) -> int:
    if l > h:
        return -1
    else:
        mid = (l + h)//2
        if nums[mid] == target:
            return mid
        elif nums[mid] > target:
            h = mid - 1
            return binarySearchRecursive(nums, target, l, h)
        else:
            l = mid + 1
            return binarySearchRecursive(nums, target, l, h)
```

```python
nums = []
target = 5
binarySearchRecursive(nums, target, 0, len(nums)-1)
```

```python
nums = [-1,0,3,5,9,12]
target = 9
binarySearchRecursive(nums, target, 0, len(nums)-1)
```
