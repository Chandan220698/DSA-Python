## Implement Lower and Upper Bound

```python
nums = [1, 1, 1, 2, 3, 3, 5, 6, 7, 7, 7, 9, 12, 12, 13]
```

### Lower bound
##### lower Bound -> Smallest index such that nums[i] >= target

```python
def lowerBound(nums: list[int], target) -> int:
    lb = len(nums)
    low = 0
    high = lb - 1
    while low <= high:
        mid = (low + high)//2
        if target <= nums[mid]:
            lb = mid
            high = mid - 1
        else:
            low = mid + 1
    return lb
```

```python
nums = [1, 1, 1, 2, 3, 3, 5, 6, 7, 7, 7, 9, 12, 12, 13]
lowerBound(nums, 1)
```

### Upper bound
##### Upper Bound -> Smallest index such that nums[i] > target

```python
def upperBound(nums: list[int], target) -> int:
    ub = len(nums)
    low = 0
    high = ub - 1
    while low <= high:
        mid = (low + high)//2
        if target < nums[mid]:
            ub = mid
            high = mid - 1
        else:
            low = mid + 1
    return ub
```

```python
nums = [1, 1, 1, 2, 3, 3, 5, 6, 7, 7, 7, 9, 12, 12, 13]
upperBound(nums, 1)
```

```python
nums = [1, 1, 1, 2, 3, 3, 5, 6, 7, 7, 7, 9, 12, 12, 13]
upperBound(nums, 7)
```

```python
nums = [1, 1, 1, 2, 3, 3, 5, 6, 7, 7, 7, 9, 12, 12, 13]
upperBound(nums, 15)
```
