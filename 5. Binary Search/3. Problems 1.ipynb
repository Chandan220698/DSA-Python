## Leetcode 35: Search Insert Position
Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

You must write an algorithm with O(log n) runtime complexity.


```python
def searchInsert(nums: list[int], target: int) -> int:
        lb = len(nums)
        low = 0
        high = lb - 1
        while low <= high:
            mid = (low + high)//2
            if target <= nums[mid]:
                lb = mid
                high = mid - 1
            else:
                low = mid + 1
        return lb
```

## Floor and Ceil

```python
def floorAndCeil(nums: list[int], target: int) -> list[int]:
    n = len(nums)
    low = 0
    high = n - 1
    ceil = -1
    floor = -1
    while low <= high:
        mid = (low + high)//2
        if target == nums[mid]:
            return [target, target]
        elif target < nums[mid]:
            ceil = nums[mid]
            high = mid - 1
        else:
            floor = nums[mid]
            low = mid + 1
    return [floor, ceil]
```

```python
nums = [1, 1, 2, 2, 4, 6, 8, 10, 11, 11, 12]
floorAndCeil(nums, 0)
```
