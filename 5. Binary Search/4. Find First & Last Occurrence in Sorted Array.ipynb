## Leetcode 34: Find First & Last Occurrence in Sorted Array

```python
def searchRangeLB(nums: list[int], target: int) -> list[int]:
    n = len(nums)
    lb = -1
    i = 0
    j = n-1

    while i <= j:
        mid = (i + j)//2
        if target <= nums[mid]:
            if target == nums[mid]:
                lb = mid
            j = mid - 1
        elif target > nums[mid]:
            i = mid + 1
    return lb
```

```python
nums = [5,7,7,8,8,10]
target = 6
searchRange(nums, target)
```

```python
searchRange([], 0)
```

```python
def searchRangeUB(nums: list[int], target: int) -> list[int]:
    n = len(nums)
    ub = -1
    i = 0
    j = n-1

    while i <= j:
        mid = (i + j)//2
        if target < nums[mid]:
            j = mid - 1
        elif target >= nums[mid]:
            i = mid + 1
            if target == nums[mid]:
                ub = mid
    return ub
```

```python
nums = [5,7,7,8,8,10]
target = 6
searchRangeUB(nums, target)
```

```python
from typing import List

class Solution:
    def searchRangeLB(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)
        lb = -1
        i = 0
        j = n-1

        while i <= j:
            mid = (i + j)//2
            if target <= nums[mid]:
                if target == nums[mid]:
                    lb = mid
                j = mid - 1
            elif target > nums[mid]:
                i = mid + 1
        return lb
    def searchRangeUB(self, nums: List[int], target: int) -> List[int]:
        n = len(nums)
        ub = -1
        i = 0
        j = n-1

        while i <= j:
            mid = (i + j)//2
            if target < nums[mid]:
                j = mid - 1
            elif target >= nums[mid]:
                i = mid + 1
                if target == nums[mid]:
                    ub = mid
        return ub
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        lb_idx = self.searchRangeLB(nums, target)
        if lb_idx == -1:
            return [-1, -1]
        ub_idx = self.searchRangeUB(nums, target)
        return [lb_idx, ub_idx]
```

```python
search = Solution()
search.searchRange(nums = [5,7,7,8,8,10], target = 8)
```

```python
search.searchRange(nums = [5,7,7,8,8,10], target = 6)
```

```python
search.searchRange(nums = [], target = 0)
```
