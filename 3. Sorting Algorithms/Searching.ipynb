## Insertion Sort

```python
def insertionSort(arr: list):
    for i in range(1, len(arr)):
        j = i-1
        while (j >= 0) and (arr[j] > arr[j+1]):
            temp = arr[j+1]
            arr[j+1] = arr[j]
            arr[j] = temp
            j -= 1
    return arr
```

```python
insertionSort([1, 2, 3, 4, 5])
```

```python
insertionSort([5, 4, 3, 2, 1])
```

```python
insertionSort([5, 1, 1, 2, 0, 0])
```

## Selection Sort

```python
def selectionSort(arr: list):
    for i in range(len(arr)):
        min_idx = i
        for j in range(i+1, len(arr)):
            if arr[j] < arr[min_idx]:
                min_idx = j
        ## swap
        temp = arr[i]
        arr[i] = arr[min_idx]
        arr[min_idx] = temp
        
    return arr
```

```python
insertionSort([1, 2, 3, 4, 5])
```

```python
insertionSort([5, 4, 3, 2, 1])
```

```python
insertionSort([5, 1, 1, 2, 0, 0])
```

## Merge Sort

```python
def mergeArrays(l_arr: list, r_arr: list):
    
    result = list()
    i,j = 0, 0
    n = len(l_arr)
    m = len(r_arr)

    while (i < n) and (j < m):
        if l_arr[i] <= r_arr[j]:
            result.append(l_arr[i])
            i += 1
        else:
            result.append(r_arr[i])
            j += 1
    if i < n:
        while i < n:
            result.append(l_arr[i])
            i += 1
    if j < m:
        while j < m:
            result.append(r_arr[j])
            j += 1
    return result
```

```python
mergeArrays([1, 2, 3, 4], [1, 1, 3, 4, 5, 6, 7, 10])
```
